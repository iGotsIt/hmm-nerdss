## Observing if we can fit a Gaussian Mixture Model with 3 components to each of the intensity distributions

```python
gmm = GaussianMixture(n_components=3, covariance_type='diag')
gmm.fit(green.reshape(-1, 1))
plt.hist(green, bins=50, alpha=0.5, label='Green', color='green')
```

```python
from scipy.stats import norm

x = np.linspace(green.min(), green.max(), 1000)

gmm = GaussianMixture(n_components=3, covariance_type='diag')
gmm.fit(green.reshape(-1, 1))

# density=True puts the histogram on the same scale as the PDFs
plt.hist(green, bins=50, density=True, alpha=0.4, color='green', label='Green data')

means   = gmm.means_.ravel()
sigmas  = np.sqrt(gmm.covariances_.ravel())
weights = gmm.weights_

# sort largest -> smallest mean, and reorder sigmas/weights to match
order   = np.argsort(means)[::-1]
means   = means[order]
sigmas  = sigmas[order]
weights = weights[order]

total = np.zeros_like(x)
for i, (mu, sigma, w) in enumerate(zip(means, sigmas, weights)):
    component = w * norm.pdf(x, mu, sigma)
    total += component
    plt.plot(x, component, linewidth=2, label=f'Component {i+1} (μ={mu:.0f}, σ={sigma:.0f}, w={w:.2f})')

plt.plot(x, total, 'k--', linewidth=2, label='Mixture sum')

plt.xlabel('Intensity')
plt.ylabel('Density')
plt.legend()
plt.show()

print("Means:", means)
print("Sigmas:", sigmas)
print("Weights:", weights)
```

```python
from scipy.stats import norm

gmm = GaussianMixture(n_components=3, covariance_type='diag')
gmm.fit(red.reshape(-1, 1))
x = np.linspace(red.min(), red.max(), 1000)

# density=True puts the histogram on the same scale as the PDFs
plt.hist(red, bins=50, density=True, alpha=0.4, color='red', label='Red data')

# pull out the fitted parameters (each is shape (n_components, ...) )
means   = gmm.means_.ravel()
sigmas  = np.sqrt(gmm.covariances_.ravel())
weights = gmm.weights_

# sort largest -> smallest mean, and reorder sigmas/weights to match
order   = np.argsort(means)[::-1]
means   = means[order]
sigmas  = sigmas[order]
weights = weights[order]

total = np.zeros_like(x)
for i, (mu, sigma, w) in enumerate(zip(means, sigmas, weights)):
    component = w * norm.pdf(x, mu, sigma)
    total += component
    plt.plot(x, component, linewidth=2, label=f'Component {i+1} (μ={mu:.0f}, w={w:.2f})')

# the full mixture (sum of all components)
plt.plot(x, total, 'k--', linewidth=2, label='Mixture sum')

plt.xlabel('Intensity')
plt.ylabel('Density')
plt.legend()
plt.show()

print("Means:", means)
print("Sigmas:", sigmas)
print("Weights:", weights)
```
