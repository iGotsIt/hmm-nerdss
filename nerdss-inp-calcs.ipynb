```python
import numpy as np
```

```python
def time_step(da, db, na, v, sigma):
    D = da + db
    rho = na / v
    return (1 / (56 * (D))) * (
        (3 / (4 * np.pi * rho) + sigma**3) ** (1 / 3) - sigma
    ) ** 2
```

```python
ts = time_step(0, 1.5*(1000 * 1000), 240, 800000, 0.001) 
print(f"Minimum time step: {ts:.2e} seconds")
```
